```python
import torch
import math
```

```python
num_vectors = 10
num_trials = 100

for dk in [2,5,10,15,200,500, 768,12288]:
    S_before_dk_avgs = []
    S_before_dk_vars = []
    
    S_after_dk_avgs = []
    S_after_dk_vars  = []
    for i in range(100):
        Q = torch.randn(num_vectors, dk)
        K = torch.randn(num_vectors, dk)
        
        S = Q@K.T    
        S_before_dk_avg = S.mean()
        S_before_dk_avgs.append(S_before_dk_avg)
        S_before_dk_var = S.var(unbiased=False)
        S_before_dk_vars.append(S_before_dk_var)
    
        S_dk = Q@K.T/math.sqrt(dk)    
        S_after_dk_avg = S.mean()
        S_after_dk_avgs.append(S_after_dk_avg)
        S_after_dk_var = S_dk.var(unbiased=False)
        S_after_dk_vars.append(S_after_dk_var)
    
    S_before_dk_avgs_mean = torch.tensor(S_before_dk_avgs).mean()
    S_before_dk_vars_mean = torch.tensor(S_before_dk_vars).mean()
    
    S_after_dk_avgs_mean = torch.tensor(S_after_dk_avgs).mean()
    S_after_dk_vars_mean = torch.tensor(S_after_dk_vars).mean()
    
    print("\n Embedding dim: dk=", dk)
    print("S_before_dk_averages(mean)  =", S_before_dk_avgs_mean)
    print("S_before_dk_variances(mean) =", S_before_dk_vars_mean)
    print("S_after_dk_averages(mean)   =", S_after_dk_avgs_mean)
    print("S_after_dk_variances(mean)  =", S_after_dk_vars_mean)
```

```python
import torch

# Settings
n = 3           # number of tokens (rows)
trials = 10000  # repeat many times for stable estimate

for d_k in [2, 4, 8, 16, 32]:
    vars_before = []
    vars_after = []

    for _ in range(trials):
        # Q, K with mean 0, var 1
        Q = torch.randn(n, d_k)
        K = torch.randn(n, d_k)

        # Compute attention scores
        S = Q @ K.T  # shape (n, n)

        # Record variance across all dot products
        vars_before.append(S.var(unbiased=False).item())

        # Now apply scaling by sqrt(d_k)
        S_scaled = S / (d_k ** 0.5)
        vars_after.append(S_scaled.var(unbiased=False).item())

    # Average variances over all trials
    print(f"d_k={d_k:2d} | Var(before scaling): {torch.tensor(vars_before).mean():.3f} | Var(after scaling): {torch.tensor(vars_after).mean():.3f}")
```
