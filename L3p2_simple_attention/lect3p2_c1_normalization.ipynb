```python
## Various Normlaizations
This code deals with various normalizations and why softmax is the best
- L1 Normalization
- L2 Normalization
- Softmax Normalization

In the bar charts notice the following
- the Sum is 1 or not
- the Magnitude is it 1 or not
- are all the normalized values positive
- is the highest value enlarged the most (like in softmax)
```

```python
import torch
import numpy as np
import torch.nn.functional as F
```

```python
# Dot product of the word "Big" with all other vectors
# @ Does matrix multiplicaation . This is much faster than using for loops
# Note inputs[1].T i.e. transpose not necessary. Pytorch can infer this automatically
dot_products = torch.tensor([1.,-2.,3.,4.,5.,6.])
print("dot_products:", dot_products)
print("dot_products.sum():",dot_products.sum())
```

### L1 Normalization
🧠 What this means
Notice that the sum is 1.
Magnitude is not 1 though.

```python
# Scale the dot_product using L1 norm
dot_products_l1 = F.normalize(dot_products, p=1, dim=0)
dot_products_l1_magnitude = torch.norm(dot_products_l1, p=2)
print("dot_products:", dot_products_l1)
print("dot_products_l1.sum():",dot_products_l1.sum())
print("dot_products_l1_magnitude:",dot_products_l1_magnitude)
```

### L2 Normalization
🧠 What this means
Notice that the sum is NOT 1.
Magnitude is 1.

```python
# Scale the dot_product using L2 norm
dot_products_l2 = F.normalize(dot_products, p=2, dim=0)
dot_products_l2_magnitude = torch.norm(dot_products_l2, p=2)
print("dot_products:", dot_products_l2)
print("dot_products_l2.sum():",dot_products_l2.sum())
print("dot_products_l2_magnitude:",dot_products_l2_magnitude)
```

## Softmax Normalization
Notice that the sum is 1.
Magnitude is not 1 though.

```python
dot_products_softmax = F.softmax(dot_products, dim=0)
dot_products_softmax_magnitude = torch.norm(dot_products_softmax, p=2)
print("dot_products:", dot_products_softmax)
print("dot_products_softmax.sum():",dot_products_softmax.sum())
print("dot_products_softmax_magnitude:",dot_products_softmax_magnitude)
```

### Bar Chart Visualization

```python
import matplotlib.pyplot as plt

# Labels
# Words
words = ['Dream', 'big', 'and', 'work', 'for', 'it']
labels = words
x_pos = np.arange(len(labels))

# Width of bars
width = 0.2

# Plot
plt.figure(figsize=(10,6))
plt.bar(x_pos - width,  dot_products,         width=width, label='Original')
plt.bar(x_pos,          dot_products_l1,      width=width, label='L1 Normalized')
plt.bar(x_pos + width,  dot_products_l2,      width=width, label='L2 Normalized')
plt.bar(x_pos + 2*width,dot_products_softmax, width=width, label='Softmax Normalized')

plt.xticks(x_pos + width/2, labels)
plt.ylabel('Value')
plt.title('Comparison of Normalizations (Bar Chart)')
plt.legend()
plt.grid(axis='y')
plt.show()
```

## Various Normalizations
Normalization of the following vector
dot_products = torch.tensor([1.,-2.,3.,4.,5.,6.])
- Note that L1 and L2 Norm can result in negative values (if the original vector was negative)
- For softmax and L1 the sum is 1. For L2 the magnitude is 1
- Softmax normalization  is positive always: results in positive values even for negative numbers. The absence of negative values makes softmax ideal for deep learning 
- Softmax suppresses the smaller values and magnifies the largest value quite a bit. This makes softmax ideal for classification. However when softmax is used in the deep learning layers the inputs (before normalization) , the inputs variation should not be extreme to prevent vanishing gradients(due to small inputs) and exploding gradients (due to large inputs). More on this in Lect 4p1

```python
# Plot
plt.figure(figsize=(10,6))
plt.bar(x_pos,          dot_products_l1,      width=width, label='L1 Normalized')
plt.bar(x_pos + width,  dot_products_l2,      width=width, label='L2 Normalized')
plt.bar(x_pos + 2*width,dot_products_softmax, width=width, label='Softmax Normalized')

plt.xticks(x_pos + width/2, labels)
plt.ylabel('Value')
plt.title('Comparison of Normalizations (Bar Chart)')
plt.legend()
plt.grid(axis='y')
plt.show()
```
